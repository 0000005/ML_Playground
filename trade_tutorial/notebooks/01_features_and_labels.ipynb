## Day 3：样本是一行可学习的数据

目标：创建一张 2 个日期、3 只股票的小表，并指出哪些列是特征，哪一列是标签。

```python
feature_cols = ['pe', 'momentum_20d']
label_col = 'future_return_5d'

X = df[feature_cols]
y = df[label_col]

X, y
```

## Day 5：回归、分类、排序

目标：用 5 只股票的预测收益，练习排序。

```python
pred_df = pd.DataFrame({
    'code': ['A', 'B', 'C', 'D', 'E'],
    'pred_return': [0.05, 0.02, -0.01, 0.03, 0.00]
})

pred_df['rank'] = pred_df['pred_return'].rank(ascending=False, method='first')
pred_df.sort_values('rank')
```

```python
top_2 = pred_df.sort_values('rank').head(2)
top_2
```
