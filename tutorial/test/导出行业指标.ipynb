聚宽 PIT 数据导出需求

  文件格式

  - 单个 CSV 文件，包含所有股票
  - UTF-8 编码，逗号分隔

  字段结构

  ┌────────┬───────┬────────────────────────────────────────────────────────────┬────────────┐
  │  字段  │ 类型  │                            说明                            │    示例    │
  ├────────┼───────┼────────────────────────────────────────────────────────────┼────────────┤
  │ symbol │ str   │ 股票代码，格式 sh600519 / sz000001                         │ sh600519   │
  ├────────┼───────┼────────────────────────────────────────────────────────────┼────────────┤
  │ date   │ str   │ 披露日，即财报实际对外公布的日期                           │ 2021-04-28 │
  ├────────┼───────┼────────────────────────────────────────────────────────────┼────────────┤
  │ period │ str   │ 报告期，即归属的截止日期                               │ 2021-03-31 │
  ├────────┼───────┼────────────────────────────────────────────────────────────┼────────────┤
  │ value  │ float │ 指标值                                                  │ 801010.0     │
  ├────────┼───────┼────────────────────────────────────────────────────────────┼────────────┤
  │ field  │ str   │ 指标名称，取值范围见下表                                   │ industry_code      │
  └────────┴───────┴────────────────────────────────────────────────────────────┴────────────┘

  需要导出的指标：
  - 申万行业分类,值就是整数（1=银行, 2=电子, ...）

  数据说明

  1. 每条记录 = 一只股票 × 一个报告期 × 一个指标。同一只股票同一报告期可以有多个不同 field 的记录
  2. date 必须是实际披露日，不能用报告期推算。如果聚宽没有提供披露日，请留空
  3. period 统一用日期格式，如一季报 2021-03-31，半年报 2021-06-30，三季报 2021-09-30，年报 2021-12-31

### api调用介绍
get_industry  查询股票所属行业

```python
get_industry(security, date=None)
```

**参数**

- security：标的代码。类型为字符串，形式如"000001.XSHE"；或为包含标的代码字符串的列表，形如["000001.XSHE", "000002.XSHE"]
- date：查询的日期。类型为字符串，形如"2018-06-01"或"2018-06-01 09:00:00"；或为datetime.datetime对象和datetime.date。*注意传入对象的时分秒将被忽略*。默认值为None，研究中默认值为当天，回测中默认值会随着回测日期变化而变化, 等于context.current_dt。

一个dict， key是标的代码。

````python
# 导出融合股票池股票在各季度末的申万一级行业归属
# 输入: hybrid_finance_pool_input_20090101_20260601.csv (由"导出财务指标.ipynb"生成)
# 输出格式: symbol,date,period,value,field
# - symbol: QLib/QLab 风格代码，如 sh600519 / sz000001
# - date: 行业归属没有财报披露日，留空
# - period: 查询行业归属所用的季度末日期
# - value: 申万一级行业原始代码，如 801740
# - field: 固定为 industry_code

from jqdata import *
import datetime as dt
import gc
import pandas as pd

try:
    from tqdm.notebook import tqdm
except Exception:
    try:
        from tqdm import tqdm
    except Exception:
        def tqdm(x, **kwargs):
            return x

INPUT_FILE = 'hybrid_finance_pool_input_20090101_20260601.csv'
OUTPUT_FILE = 'hongli50_industry_20100101_202605.csv'
START_DATE = '2009-01-01'
END_DATE = '2026-05-31'
BATCH_SIZE = 100


def to_qlib_symbol(jq_code):
    code, exchange = jq_code.split('.')
    prefix = 'sh' if exchange == 'XSHG' else 'sz'
    return prefix + code


def quarter_ends(start_date, end_date):
    start = pd.Timestamp(start_date)
    end = pd.Timestamp(end_date)
    dates = pd.date_range(start=start, end=end, freq='Q')
    return [d.strftime('%Y-%m-%d') for d in dates]


def chunks(seq, size):
    for i in range(0, len(seq), size):
        yield seq[i:i + size]


def write_csv_text(path, text, append=True):
    try:
        write_file(path, text, append=append)
    except Exception:
        mode = 'a' if append else 'w'
        with open(path, mode, encoding='utf-8', newline='') as f:
            f.write(text)


# 读取融合股票池输入 CSV，直接提取去重的 jq_code 列
pool_df = pd.read_csv(INPUT_FILE)
if 'jq_code' not in pool_df.columns:
    raise ValueError('输入 CSV 缺少 jq_code 列。请确认文件由"导出财务指标.ipynb"生成。')

stocks = sorted(pool_df['jq_code'].dropna().unique().tolist())

periods = quarter_ends(START_DATE, END_DATE)
print('股票数:', len(stocks), '季度数:', len(periods), '预计行数:', len(stocks) * len(periods))

write_csv_text(OUTPUT_FILE, 'symbol,date,period,value,field\n', append=False)

written_rows = 0
for period in tqdm(periods, desc='导出季度行业归属'):
    out_lines = []
    for batch in chunks(stocks, BATCH_SIZE):
        industry_result = get_industry(security=batch, date=period)
        for jq_code in batch:
            sw_l1 = industry_result.get(jq_code, {}).get('sw_l1', {})
            value = sw_l1.get('industry_code', '')
            out_lines.append('%s,,%s,%s,industry_code\n' % (to_qlib_symbol(jq_code), period, value))
        if len(out_lines) >= 5000:
            write_csv_text(OUTPUT_FILE, ''.join(out_lines), append=True)
            written_rows += len(out_lines)
            out_lines = []
            gc.collect()
    if out_lines:
        write_csv_text(OUTPUT_FILE, ''.join(out_lines), append=True)
        written_rows += len(out_lines)
        out_lines = []
        gc.collect()

print('完成:', OUTPUT_FILE, '写入行数:', written_rows)
````
